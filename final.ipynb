```python
# =====================================================================
# FORCE INSTALL VIA ACTIVE KERNEL EXECUTABLE PATH
# =====================================================================
import sys
import os

print("📦 Step 1: Forcing installation directly into the active kernel path...")
# This targets the exact python binary executing this cell
!{sys.executable} -m pip install noisereduce librosa soundfile --quiet

print("\n🎙️ Step 2: Running the audio noise reduction engine...")
import librosa
import soundfile as sf
import noisereduce as nr

input_file = "/content/Leeroy Jenkins HD 1080p - J Jonah Jameson (youtube).mp3"
output_dir = "/content/deepfilter_outputs"
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, "clean_speech_final.wav")

if not os.path.exists(input_file):
    print(f"❌ Error: Please verify your file is uploaded to the left panel as exactly:")
    print(f"   {input_file}")
else:
    print(f"⚡ Loading target audio track: {input_file}")
    audio_data, sample_rate = librosa.load(input_file, sr=None)

    print("🚀 Computing non-stationary noise reduction gates...")
    # Clean up the background noise completely
    reduced_noise_audio = nr.reduce_noise(y=audio_data, sr=sample_rate, stationary=False)

    print(f"💾 Saving clean wave track directly to disk: {output_file}")
    sf.write(output_file, reduced_noise_audio, sample_rate)

    print(f"\n✨ SUCCESS! Your file has been processed cleanly.")
    print(f"📂 Open your left sidebar file pane, hit the 'Refresh' icon, and download from:")
    print(f"   {output_file}")
```
